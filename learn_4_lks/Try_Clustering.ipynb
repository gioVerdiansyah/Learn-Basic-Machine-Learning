```python
clean_df = df.dropna(axis=0)
clean_df.isna().sum().to_frame().T.style.set_properties(**style_table)
```

```python
clean_df["Dt_Customer"] = pd.to_datetime(clean_df["Dt_Customer"], format="%d-%m-%Y", errors="coerce")

dates = []

for i in clean_df["Dt_Customer"]:
  dt = i.date()
  dates.append(dt)

print(f"NaT sum: {clean_df['Dt_Customer'].isna().sum()}")

print(f"Newest customer enrollment: {min(dates)}")
print(f"Last customer enrollment: {max(dates)}")
```

```python
import numpy as np

days = []
newest = max(dates)

for i in dates:
  dy = newest - i
  days.append(dy)

clean_df["Customer Loyalty"] = days
clean_df["Customer Loyalty"] = clean_df["Customer Loyalty"].dt.days
clean_df["Age"] = 2025 - clean_df["Year_Birth"]
clean_df["Spent"] = clean_df['MntWines'] +	clean_df['MntFruits'] +	clean_df['MntMeatProducts'] +	clean_df['MntFishProducts'] +	clean_df['MntSweetProducts'] +	clean_df['MntGoldProds']
clean_df["Children"] = clean_df["Kidhome"] + clean_df["Teenhome"]
clean_df["Parental Status"] = np.where(clean_df["Children"] > 0, 1, 0)

clean_df["Marital_Status"].replace({"Married":"Couple", "Together":"Couple", "Absurd":"Alone", "Widow":"Alone", "YOLO":"Alone", "Divorced":"Alone", "Single":"Alone"}, inplace = True)
clean_df["Education"].replace({"Basic":"Undergraduate","2n Cycle":"Undergraduate", "Graduation":"Graduate", "Master":"Postgraduate", "PhD":"Postgraduate"}, inplace = True)
clean_df["Total Promo"] =  clean_df["AcceptedCmp1"]+ clean_df["AcceptedCmp2"]+ clean_df["AcceptedCmp3"]+ clean_df["AcceptedCmp4"]+ clean_df["AcceptedCmp5"]
clean_df.rename(columns={"Marital_Status": "Marital Status","MntWines": "Wines","MntFruits":"Fruits","MntMeatProducts":"Meat","MntFishProducts":"Fish","MntSweetProducts":"Sweets","MntGoldProds":"Gold","NumWebPurchases": "Web","NumCatalogPurchases":"Catalog","NumStorePurchases":"Store","NumDealsPurchases":"Discount Purchases"}, inplace = True)
clean_df = clean_df[["ID", "Age","Education","Marital Status","Parental Status","Children","Kidhome","Teenhome","Income","Spent","Customer Loyalty","Recency","Wines","Fruits","Meat","Fish","Sweets","Gold","Web","Catalog","Store","Discount Purchases","Total Promo","NumWebVisitsMonth"]]

clean_df = clean_df[clean_df["Age"] < 90]
clean_df = clean_df[clean_df["Income"] < 600000]
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

cat_cols = ['Education', 'Marital Status']
for col in cat_cols:
  clean_df[col] = le.fit_transform(clean_df[col])
```

```python
from sklearn.preprocessing import StandardScaler

data = clean_df.copy()

scaler = StandardScaler()
data = scaler.fit_transform(data)
scaled_data = pd.DataFrame(data, columns=clean_df.columns)
scaled_data.head().style.set_properties(**style_table)
```

```python
one = clean_df.loc[0:1]
one
```

```python
from sklearn.decomposition import PCA

scaled_dataC = scaled_data.copy()
pca = PCA(n_components=3, random_state=42)
scaled_dataC = pca.fit_transform(scaled_dataC)
pca_data = pd.DataFrame(scaled_dataC, columns=["col1", "col2", "col3"])

pca_data.head(15).T.style.set_properties(**style_table)
```

```python
pca_one = pca_data.iloc[1:2]
pca_one
```

```python
import plotly.graph_objects as go

x = pca_data["col1"]
y = pca_data["col2"]
z = pca_data["col3"]

print(f"Let's have a look on the dataset after dimensionality reduction :")

fig = go.Figure(data = [go.Scatter3d(x=x, y=y, z=z,mode='markers',
                marker = dict(size=10, color="#682F2F", opacity=0.8))])
fig.update_layout(title=dict(text= "Dataset After Diemsionality Reduction[3D Projection]",
                             y=0.97,x=0.5,xanchor= "center",yanchor= "top",font_color="black"),
                  scene=dict(xaxis = dict(title="col1",backgroundcolor="#D4C8BA",showbackground=True,showgrid=True, gridwidth=2, gridcolor='white',zerolinecolor="white"),
                             yaxis = dict(title="col2",backgroundcolor="#D4C8BA",showbackground=True,showgrid=True, gridwidth=2, gridcolor='white',zerolinecolor="white"),
                             zaxis = dict(title="col3",backgroundcolor="#D4C8BA",showbackground=True,showgrid=True, gridwidth=2, gridcolor='white',zerolinecolor="white")),
                  margin=dict(l=0, r=0, b=0, t=0),paper_bgcolor="#D4C8BA")
fig.show()
```

```python
# KElbow to determining cluster
from yellowbrick.cluster import KElbowVisualizer
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
import seaborn as sns

_, axes = plt.subplots(figsize=(18,8))

elbow = KElbowVisualizer(KMeans(), k=10, timings=False, locate_elbow=True, size=(1260, 460))
elbow.fit(pca_data)

axes.set_title("\nDistortion Score Elbow For KMeans Clustering\n",fontsize=25)
axes.set_xlabel("\nK",fontsize=20)
axes.set_ylabel("\nDistortion Score",fontsize=20)

sns.despine(left=True, bottom=True)
plt.show()
```

```python
km = KMeans(n_clusters=4, random_state=42)
y_km = km.fit_predict(pca_data)

pca_data["Clusters"] = y_km
clean_df["Clusters"] = y_km
```

```python
print("Let's have a look on the customers quantity distribution after clustering :")

plt.subplots(figsize=(20, 8))
p = sns.countplot(x=clean_df["Clusters"],palette=["#B9C0C9","#682F2F", "#9F8A78","#F3AB60"], saturation=1,edgecolor = "#1c1c1c", linewidth = 4)
p.axes.set_yscale("linear")
p.axes.set_title("\nCustomer's Clusters\n",fontsize=25)
p.axes.set_ylabel("Count",fontsize=20)
p.axes.set_xlabel("\nCluster",fontsize=20)
p.axes.set_xticklabels(p.get_xticklabels(),rotation = 0)
for container in p.containers:
    p.bar_label(container,label_type="center",padding=6,size=30,color="black",rotation=0,
    bbox={"boxstyle": "round", "pad": 0.4, "facecolor": "orange", "edgecolor": "white", "linewidth" : 4, "alpha": 1})


sns.despine(left=True, bottom=True)
plt.show()
```
