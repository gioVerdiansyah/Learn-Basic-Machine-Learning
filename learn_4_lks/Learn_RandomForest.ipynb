```python
import numpy as np
import pandas as pd

# Generate dataset
np.random.seed(42)
X = np.random.rand(20, 3)  # 20 samples, 3 features
y = np.random.choice([0, 1], size=20)  # Binary classification

data = pd.DataFrame(X, columns=["Feature1", "Feature2", "Feature3"])
data["Label"] = y

data.head()
```

```python
# Example split at index 0 (Feature1 < threshold)
threshold = 0.3
left_indices = data["Feature1"] < threshold
right_indices = ~left_indices

y_left = y[left_indices]
y_right = y[right_indices]

y_left
```
