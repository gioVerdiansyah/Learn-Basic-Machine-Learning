```python
from flask import Flask, render_template, request
import pickle
import numpy as np
```

```python
app = Flask(__name__, template_folder="/content/drive/MyDrive/learn_ML_LKS/exercise/collaboration/template")

model = pickle.load(open("/content/drive/MyDrive/learn_ML_LKS/exercise/models/forecast_model.pkl", 'rb'))

@app.route('/')
def home():
  return "Hello wordl"

@app.route('/predict', methods=['POST'])
def predict():
  if request.method == "POST":
    input = [x for x in request.from_values()]
    final_output = [np.array(input)]
    prediction = model.predict(final_output)[0]
    return render_template('index.html', output=format(prediction.round(2)))

if __name__ == '__main__':
  app.run()
```
