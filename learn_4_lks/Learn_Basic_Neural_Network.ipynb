Membangun Neaural Network dengan Numpy

Contoh saya memiliki vector dan 2 bobot vector lainnya. Tujuannya adalah mencari bobot mana yang paling mirip dengan input, berdasarkan arah dari input, contohnya seperti ini:

![Example img](https://files.realpython.com/media/three_vectors_2d.9d220456ff49.png)

weights_2 lebih mirip dengan input vector dari arah titik yang sama besaran, jadi bagaimana
weights_2 is more similar to the input vector since it's pointing in the same direction and the magnitude is also similar. So how do you figure out which vectors are similar using Python?

First, define the 3 vectors, 1 for input and 2 other for the weight. Then you compute how smiliar input_vector and weight_1 are. To do that:

1. Multiply the first index of input_vector by the first index of weights_1
2. Multiply the second index of indput_vector by the second index of weights_2
3. Sum the results of both multiplications.

```python
input_vec = [1.83, 1.25]
weight_1 = [1.25, 0]
weight_2 = [2.27, 0.32]

# Computing the dot product
first_index_mult = input_vec[0] * weight_1[0]
second_index_mult = input_vec[1] * weight_1[1]
dot_product_1 = first_index_mult + second_index_mult

print(f"The dot product is: {dot_product_1}")
```

Use with numpy

```python
import numpy as np

dot_product_1 = np.dot(input_vec, weight_1)

print(f"The dot product 1 is: {dot_product_1}")

dot_product_2 = np.dot(input_vec, weight_2)

print(f"The dot product 1 is: {dot_product_2}")
```

# Make Prediction<br>
So far, you've seen that the only two operations used inside the neural network were the dot product and a sum. Both are linear operations.

If you add more layers but keep using only linear operations, then adding more layers would have no effect because each layer will always have some correlation with the input of the previous layer.

You can achieve this behavior by using nonlinear functions. These nonlinear functions are called activation functions.

activation functions is determine the neuron is on or off by weighted sum from input

There are many types of activation functions like :

- __ReLU__ func to convert all negative numbers to zero
- __Sigmoid__ func to limit output to range 0 and 1, for binary classification
- __Softmax__ change values to probabilities, for categorical classification

The network we building is Sigmoid activation:
![sigmoid formula](https://files.realpython.com/media/sigmoid_function.f966c820f8c3.png)

Since the function limits the output to a range of 0 to 1, you’ll use it to predict probabilities. If the output is greater than 0.5, then you’ll say the prediction is 1. If it’s below 0.5, then you’ll say the prediction is 0. This is the flow of the computations inside the network you’re building:
![path](https://files.realpython.com/media/network_architecture.406cfcc68417.png)

```python
# Input Vector      Target
# [1.61, 1.54]      1
# [2.1, 1.4]        0
# if the output is > 0.5 = 1 | < 0.5 = 0

input_vec = np.array([1.61, 1.54])
weight_1 = np.array([1.40, -0.59])
bias = np.array([0.0])

def sigmoid(x):
  return 1 / (1 + np.exp(-x))

def make_prediction(input, weight, bias):
  layer_1 = np.dot(input, weight) + bias
  layer_2 = sigmoid(layer_1)
  return layer_2

pred = make_prediction(input_vec, weight_1, bias)

print(f"The prediction result is: {pred}")
# [0.79337657] > 0.5 = 1 | True
```

```python
input_vec = np.array([2.1, 1.4])
pred = make_prediction(input_vec, weight_1, bias)

print(f"The prediction result is: {pred}")

# The value is [0.89225648] and the correct value must < 0.5 | This model is wrong
```

# Train Neural Network<br>
Training NN first is assess the error and then adjust the weight. To adjust the weights, we'll use the __gradient descent__ and __backpropagation__ algorithms.<br>

__gradient descent__ is applied to find the direction and the rate to update the parameters.

Before making any changes in the network, you need to compute the error. That’s what you’ll do in the next section.

# Computing the Prediction Error<br>
The function used to measure the error is called __cost fucntion__, or __lost function__.<br>
We will use Mean Squared Error(MSE) as cost function, claculate MSE is 2 sptes:
1. Compute the difference between prediction and target
2. Multiply the result by itself


```python
target = 0
mse = np.square(pred - target) # pow the number in array

print(f"Prediction: {pred} | Error {mse}")
```

# Understanding How to Reduce the Error<br>
The goal is chane the weight and bias variabel so you can reduce the error.<br>

You compute the MSE by doing error = np.square(prediction - target). If you treat (prediction - target) as a single variable x, then you have error = np.square(x), which is a quadratic function. Here’s how the function looks if you plot it:
![Plot](https://files.realpython.com/media/quatratic_function.002729dea332.png)
The error is given by the y-axis. If you’re in point A and want to reduce the error toward 0, then you need to bring the x value down. we'll use __derivative__, another is use gradient descent. Gradient desent is name of algoritm to find the direction and the rate to update the network parameters.

```python
derivative = 2 * (pred - target)
print(f"The derivative is {derivative}")
```
