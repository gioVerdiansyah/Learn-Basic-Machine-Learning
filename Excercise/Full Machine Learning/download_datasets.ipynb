```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("hopesb/hr-analytics-dataset")

print("Path to dataset files:", path)
```

# Extract

```python
import shutil
import os

source_path = "/root/.cache/kagglehub/datasets/hopesb/hr-analytics-dataset/versions/2"
destination_path = "/content/datasets"

os.makedirs(destination_path, exist_ok=True)

for file_name in os.listdir(source_path):
    source_file = os.path.join(source_path, file_name)
    destination_file = os.path.join(destination_path, file_name)

    shutil.copy(source_file, destination_file)

print("Dataset berhasil dipindahkan ke:", destination_path)
print(os.listdir(destination_path))
```
