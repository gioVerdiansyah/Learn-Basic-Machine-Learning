Before that, we must know about Hyperparameter is a variable make by developer to control machine learning process like training speed, the number of layers dan neuron in neural network, batch size, etc.

Batch Loading is a training process when Neural Network update learning parameter (weight) while reading number of samples.

For example, we have 800 sample and without batch size the conclusion is in 1 epoch there is a 800 sample this mean all sample we used to updated weight.

Example we use batch size 32, the conclusion is 800/32 is 25, so in 1 epoch there is a just 25 sample we use to updated weight, this more than smaller without batch size, and of course with batch size we can save more time.

```python
import tensorflow_datasets as tfds
tfds.__version__
```

```python
(train_img, train_label), (test_img, test_label) = tfds.as_numpy(
    tfds.load(
        "mnist",
        split=["train", "test"],
        batch_size=-1,
        as_supervised=True
    )
)
```

```python
import tensorflow as tf

model = tf.keras.models.Sequential([
    tf.keras.layers.Conv2D(32, (3,3), activation="relu", input_shape=(28,28,1)),
    tf.keras.layers.MaxPool2D(2,2),
    tf.keras.layers.Conv2D(64, (3,3), activation="relu"),
    tf.keras.layers.MaxPool2D(2,2),
    tf.keras.layers.Conv2D(128, (3,3), activation="relu"),
    tf.keras.layers.MaxPool2D(2,2),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(62, activation="relu"),
    tf.keras.layers.Dense(128, activation="relu"),
    tf.keras.layers.Dense(10, activation="relu"),
])

model.compile(
    optimizer=tf.keras.optimizers.RMSprop(),
    loss=tf.keras.losses.SparseCategoricalCrossentropy(),
    metrics=["accuracy"]
)
```

```python
model.fit(train_img, train_label, batch_size=64 ,epochs=5)
```

Look at the time, if we raise the batch size it will be more fast to train, but carefully if sample to low may model is underfitting/overfitting, and may you can add more epochs

```python
model.fit(train_img, train_label, batch_size=128, epochs=5)
```
