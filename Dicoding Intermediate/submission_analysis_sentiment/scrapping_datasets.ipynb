### Scrapping Honkai Impact 3 Reviews with Google play scrapper

```python
from google_play_scraper import Sort, reviews, reviews_all
import pandas as pd

scrap_review = reviews_all(
    "com.miHoYo.bh3oversea",
    lang="id",
    country="id",
    sort=Sort.MOST_RELEVANT,
    # count=15000
)
```

### Saving to csv

```python
import os

datasets_path = "./datasets/honkai_impact_reviews.csv"
if not os.path.exists(datasets_path):
    os.makedirs(datasets_path)

df = pd.DataFrame(scrap_review)
df.to_csv(datasets_path)

df.shape
```
