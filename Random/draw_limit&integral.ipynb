```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Definisi fungsi
def f(x):
    return (x**3 - 3) / (x - 1)


# Rentang x (hindari x = 1 karena penyebut = 0)
x = np.linspace(0, 3, 400)
x = x[x != 1]  # hilangkan nilai x = 1

# Hitung nilai fungsi
y = f(x)

# Titik-titik mendekati limit di x = 2
x_points = np.array([1.5, 1.8, 1.99, 2, 2.01, 2.1, 2.2])
y_points = f(x_points)

# Plot fungsi
plt.figure(figsize=(8, 6))
plt.plot(x, y, label=r"$f(x)=\frac{x^3-3}{x-1}$", color="blue")

# Plot titik-titik tabel
plt.scatter(x_points, y_points, color="red", zorder=5, label="Pendekatan x→2")

# Garis bantu untuk limit di x=2
plt.axvline(2, color="green", linestyle="--", label="x=2")
plt.axhline(5, color="orange", linestyle="--", label="Limit ≈ 5")

# Hiasan sumbu Cartesius
plt.axhline(0, color="black", linewidth=1)
plt.axvline(0, color="black", linewidth=1)

# Label dan judul
plt.title("Visualisasi Limit f(x) = (x³ - 3)/(x - 1) saat x → 2")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid(True)

# Tampilkan
plt.show()
```
